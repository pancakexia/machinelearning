# **Imports**

```python
import matplotlib.pyplot as plt
import numpy as np
import time
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import torchvision
```

# **Parameter initialization and data preparation**

```python
# Parameters
batch_size = 8
num_epochs = 10
device = 'cuda:0'
num_classes = 10

# Load dataset
transform = torchvision.transforms.Compose(
    [torchvision.transforms.ToTensor(),
     torchvision.transforms.Normalize((0.5), (0.5))])

trainset = torchvision.datasets.CIFAR10(root='./data', train=True,
                                        download=True, transform=transform)
valset = torchvision.datasets.CIFAR10(root='./data', train=False,
                                       download=True, transform=transform)
classes = ('plane', 'car', 'bird', 'cat',
           'deer', 'dog', 'frog', 'horse', 'ship', 'truck')

print(trainset)

# Create dataloaders
trainloader = torch.utils.data.DataLoader(trainset, batch_size=batch_size,
                                          shuffle=True)
valloader = torch.utils.data.DataLoader(valset, batch_size=batch_size,
                                         shuffle=False)
```

```python
def imshow(img):
    img = img / 2 + 0.5     # unnormalize to show images correctly
    npimg = img.numpy()
    plt.imshow(np.transpose(npimg, (1, 2, 0)))
    plt.show()
```

```python
# Print some samples of dataset as a sanity check

# Get some random training images
dataiter = iter(trainloader)
example_images, example_labels = next(dataiter)
print(example_images.shape)

# Show images
imshow(torchvision.utils.make_grid(example_images))
# Print labels
print(' '.join('%5s' % classes[example_labels[j]] for j in range(batch_size)))
```

# **Define Models**

```python
class DenseNet(nn.Module):
    def __init__(self, input_features, num_classes):
        # Instantiation of layers and creation of trainable parameters
        super().__init__()
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(input_features, 1024)
        self.fc2 = nn.Linear(1024, 512)
        self.fc3 = nn.Linear(512, 256)
        self.fc4 = nn.Linear(256, 128)
        self.fc5 = nn.Linear(128, 64)
        self.fc6 = nn.Linear(64, num_classes)
        # >> Your code goes here <<
        # (You can copy the code from last lecture)

    def forward(self, x):
        # Forward pass: the computations that are done on the input -> output
        x = self.flatten(x)
        x = F.relu(self.fc1(x)) # Apply ReLU activation and assign to x
        x = F.relu(self.fc2(x)) # Apply ReLU activation and assign to x
        x = F.relu(self.fc3(x)) # Apply ReLU activation and assign to x
        x = F.relu(self.fc4(x)) # Apply ReLU activation and assign to x
        x = F.relu(self.fc5(x)) # Apply ReLU activation and assign to x
        x = self.fc6(x) # Output
        # >> Your code goes here <<
        # (You can copy the code from last lecture)
        return x
```

```python
class ConvNet(nn.Module):
    def __init__(self, input_channels, num_classes):
        super().__init__()
        self.conv1 = nn.Conv2d(input_channels, 6, 5,stride=1, padding=0)
        self.pool = nn.MaxPool2d(2,stride=2)
        self.conv2 = nn.Conv2d(6, 16, 5)
        self.pool = nn.MaxPool2d(2, stride=2)
        self.fc1 = nn.Linear(16 * 5 * 5, 120)  # Added a fully connected layer
        self.fc2 = nn.Linear(120, 84)  # Added a fully connected layer
        self.fc3 = nn.Linear(84, num_classes)  # Added a fully connected layer
        # >> Your code goes here <<

        # (You can copy the code from last lecture)

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))
        x = self.pool(F.relu(self.conv2(x)))
        x = torch.flatten(x, 1)  # Flatten the tensor
        x = F.relu(self.fc1(x))  # Pass through fully connected layers
        x = F.relu(self.fc2(x))
        x = self.fc3(x)
        # >> Your code goes here <<
        # (You can copy the code from last lecture)
        return x
```

# **Make a training loop**

```python
def compute_run_acc(logits, labels):
    _, pred = torch.max(logits.data, 1)
    return (pred == labels).sum().item()
```

```python
def count_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)
```

```python
# Instantiate model and optimizer
# RUN THIS CELL twice: once with "conv_net = False" and once "conv_net = True"

conv_net = True

if conv_net:
    model = ConvNet(3, num_classes).to(device)
    print("Convolutional model loaded")
    tr_accuracies_conv = np.zeros(num_epochs)
    val_accuracies_conv = np.zeros(num_epochs)
else:
    model = DenseNet(1024*3, num_classes).to(device)
    print("Dense model loaded")
    tr_accuracies_dense = np.zeros(num_epochs)
    val_accuracies_dense = np.zeros(num_epochs)
print("Number of trainable parameters: {}".format(count_parameters(model)))

criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=0.001, momentum=0.9)
best_val_acc = 0

start_time = time.time()

for epoch_nr in range(num_epochs):

    print("Epoch {}:".format(epoch_nr))

    # Train model
    running_loss = 0.0
    running_acc = 0.0

    # Train model
    for batch_data, batch_labels in trainloader:
      batch_data = batch_data.to(device)
      batch_labels = batch_labels.to(device)
      optimizer.zero_grad()
      logits = model(batch_data)
      loss = criterion(logits, batch_labels)
      loss.backward()
      optimizer.step()
      running_loss += loss.item()
      running_acc += compute_run_acc(logits, batch_labels)
    # >> Your code goes here <<
    # You can re-use code from previous lab session

    # Print results
    tr_acc = 100 * running_acc/len(trainloader.dataset)
    print('>> TRAIN: Epoch {} completed | tr_loss: {:.4f} | tr_acc: {:.2f}%'.format(
        epoch_nr, running_loss/len(trainloader.dataset), tr_acc))

    # Get validation results
    running_eval_loss = 0.0
    for batch_data, batch_labels in valloader:
      batch_data = batch_data.to(device)
      batch_labels = batch_labels.to(device)
      logits = model(batch_data)
      loss = criterion(logits, batch_labels)
      running_eval_loss += loss.item()
    # >> Your code goes here <<
    # You can re-use code from previous lab session
    # (beware the variablename testloader has been changed in valloader)

    val_acc = 100 * running_eval_loss/len(valloader.dataset)
    print('>> VALIDATION: Epoch {} | val_acc: {:.2f}%'.format(epoch_nr, val_acc))

    if conv_net:
        tr_accuracies_conv[epoch_nr] = tr_acc
        val_accuracies_conv[epoch_nr] = val_acc
    else:
        tr_accuracies_dense[epoch_nr] = tr_acc
        val_accuracies_dense[epoch_nr] = val_acc

    # Save model if best accuracy on validation dataset until now
    best_val_acc=0
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), 'best_model.pth')
    # >> Your code goes here <<
    # You can re-use code from previous lab session

end_time = time.time()
print('Finished Training in {:.2f} seconds'.format(end_time-start_time))
```
